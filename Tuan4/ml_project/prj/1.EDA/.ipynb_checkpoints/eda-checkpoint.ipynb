```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Đọc dữ liệu
train_df = pd.read_csv('../../data/train.csv')
test_df = pd.read_csv('../../data/test.csv')

# Log-transform biến mục tiêu (SalePrice) để giảm độ lệch
train_df['SalePrice_Log'] = np.log1p(train_df['SalePrice'])

# Lọc các đặc trưng có độ tương quan > 0.5 với SalePrice
numeric_cols = train_df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()
top_features = corr_matrix.index[abs(corr_matrix["SalePrice"]) > 0.5]

print("Các đặc trưng tương quan mạnh nhất:\n", top_features.tolist())
```
