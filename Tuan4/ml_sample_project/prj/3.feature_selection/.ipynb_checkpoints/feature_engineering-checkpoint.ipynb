```python
import pandas as pd
import numpy as np

# Đọc dữ liệu đã qua tiền xử lý
df = pd.read_csv('../../exps_/data/feature1/df_preprocessed.csv')
train_len = 1460 # Số dòng của tập train.csv ban đầu

# 1. Feature Engineering (Tạo đặc trưng mới)
# Tạo các đặc trưng mới cùng lúc để tránh PerformanceWarning
new_features = pd.DataFrame({
    # Tổng diện tích nhà
    'TotalSF': df['1stFlrSF'] + df['2ndFlrSF'] + df['TotalBsmtSF'],

    # Tổng số phòng tắm
    'TotalBath': df['FullBath'] + (0.5 * df['HalfBath']) + df['BsmtFullBath'] + (0.5 * df['BsmtHalfBath']),

    # Tuổi của căn nhà lúc bán
    'HouseAge': df['YrSold'] - df['YearBuilt'],

    # Tuổi của căn nhà kể từ lần sửa chữa gần nhất
    'RemodAge': df['YrSold'] - df['YearRemodAdd'],

    # Nhà có hồ bơi hay không
    'HasPool': (df['PoolArea'] > 0).astype(int)
})

# Thêm các đặc trưng mới vào DataFrame
df = pd.concat([df, new_features], axis=1)

# 2. Xóa các cột gốc đã được thay thế để tránh đa cộng tuyến
cols_to_drop = [
    '1stFlrSF',
    '2ndFlrSF',
    'TotalBsmtSF',
    'FullBath',
    'HalfBath',
    'BsmtFullBath',
    'BsmtHalfBath',
    'YearBuilt',
    'YearRemodAdd'
]

df.drop(cols_to_drop, axis=1, inplace=True)

# 3. Tách lại tập Train và Test
X_train = df.iloc[:train_len, :]
X_test = df.iloc[train_len:, :]

# 4. Xuất file cuối cùng cho thành viên khác sử dụng
X_train.to_excel('../../exps_/data/feature1/x_train.xlsx', index=False)
X_test.to_excel('../../exps_/data/feature1/x_test.xlsx', index=False)

# Giả sử y_train đã được lưu ở file trước hoặc có thể đọc lại từ train gốc
train_raw = pd.read_csv('../../data/train.csv')
y_train = np.log1p(train_raw['SalePrice'])

y_train.to_frame(name='SalePrice').to_excel(
    '../../exps_/data/feature1/y_train.xlsx',
    index=False
)

print("Kích thước X_train cuối cùng:", X_train.shape)
print("Kích thước X_test cuối cùng:", X_test.shape)
print("Đã lưu x_train.xlsx, y_train.xlsx, x_test.xlsx thành công vào exps_/data/feature1/")
```
