```python
import pandas as pd
import numpy as np

# 1. Cấu trúc Sheet: Baselinne (So sánh các mô hình gốc)
baseline_data = [
    [np.nan, np.nan, np.nan, np.nan, np.nan, np.nan],
    [np.nan, 'Base', 'Linear Regression', '0.0000', '0.0000', '0.0000'],
    [np.nan, np.nan, 'Random Forest', '0.0000', '0.0000', '0.0000'],
    [np.nan, np.nan, 'XGBoost', '0.0000', '0.0000', '0.0000'],
    [np.nan, np.nan, 'PyTorch MLP', '0.0000', '0.0000', '0.0000']
]
df_baseline = pd.DataFrame(baseline_data, columns=['Unnamed: 0', 'Unnamed: 1', 'Unnamed: 2', 'Train RMSE', 'Validation RMSE', 'Kaggle Score'])

# 2. Cấu trúc Sheet: Fusion (Ablation Study - Tác động của Dữ liệu & Cải tiến)
fusion_data = [
    [np.nan, 'Validation set', 'RMSE Metrics', np.nan, np.nan],
    [np.nan, np.nan, 'Train RMSE', 'Validation RMSE', 'Kaggle Score'],
    [np.nan, 'Base Model (Chỉ FillNA)', '0.0000', '0.0000', '0.0000'],
    [np.nan, 'Base + Encoding & Scaling', '0.0000', '0.0000', '0.0000'],
    [np.nan, 'Base + Feature Engineering (TotalSF...)', '0.0000', '0.0000', '0.0000'],
    [np.nan, 'Ensemble (Tối ưu mô hình ML + PyTorch)', '0.0000', '0.0000', '0.0000']
]
df_fusion = pd.DataFrame(fusion_data, columns=['Unnamed: 0', 'Unnamed: 1', 'Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'])

# 3. Xuất file Excel chuẩn
with pd.ExcelWriter('Experiment_HousePrices.xlsx', engine='openpyxl') as writer:
    df_fusion.to_excel(writer, sheet_name='Fusion', index=False, header=False)
    df_baseline.to_excel(writer, sheet_name='Baselinne', index=False, header=False)

print("Đã tạo Experiment_HousePrices.xlsx với 2 sheet 'Baselinne' và 'Fusion'!")
```
