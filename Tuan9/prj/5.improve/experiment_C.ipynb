## Thiết lập và kết quả nhóm C
Các cấu hình đã chạy với ba seed 42, 123, 2026. Đọc bảng tổng hợp mean ± standard deviation trên validation; không đọc test để chọn kiến trúc.

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(ROOT/"results/experiment_C_summary.csv")
display(df.round(8))
plt.figure(figsize=(7,4))
plt.errorbar(df.factor_value, df.val_mse_mean, yerr=df.val_mse_std, marker="o", capsize=4)
plt.xlabel("Hidden layers C"); plt.ylabel("Validation MSE"); plt.title("Experiment C"); plt.grid(alpha=.3); plt.show()
```
